# Task 3 output profile

How many unique URLs and product names are in `output/task3-*.csv`, and which values repeat.
Done by AI, the solution makes sense

```python
import csv
from collections import Counter

files = {"LT": "output/task3-lt.csv", "US": "output/task3-us.csv"}
data = {}
for locale, path in files.items():
    with open(path, encoding="utf-8") as f:
        data[locale] = list(csv.DictReader(f))
```

```python
print(f"{'':4} {'rows':>5} {'unique url':>11} {'unique name':>12} {'unique id':>10}")
for locale, rows in data.items():
    print(f"{locale:4} {len(rows):>5} "
          f"{len({r['url'] for r in rows}):>11} "
          f"{len({r['product_name'] for r in rows}):>12} "
          f"{len({r['product_id'] for r in rows}):>10}")
```

```python
# Product names that appear more than once, with the URLs they appear under.
for locale, rows in data.items():
    repeated = [n for n, c in Counter(r["product_name"] for r in rows).items() if c > 1]
    print(f"{locale}: {len(repeated)} repeated product names")
    for name in repeated:
        print(f"   {name!r}: {[r['url'] for r in rows if r['product_name'] == name]}")
```

```python
# URLs that appear more than once.
for locale, rows in data.items():
    repeated = [u for u, c in Counter(r["url"] for r in rows).items() if c > 1]
    print(f"{locale}: {len(repeated)} repeated URLs {repeated}")
```

```python
# Product names found in only one locale.
lt_names = {r["product_name"] for r in data["LT"]}
us_names = {r["product_name"] for r in data["US"]}
print("only in LT:", sorted(lt_names - us_names))
print("only in US:", sorted(us_names - lt_names))
```
